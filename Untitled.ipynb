```python
import serial, time

bt = serial.Serial("/dev/ttyPS0", 115200, timeout=1)
time.sleep(1)

bt.write(b"AT")
time.sleep(0.3)

print("RX:", bt.read(64))
```

```python
import serial, time

bt.close()
bt = serial.Serial("/dev/ttyPS0", 9600, timeout=1)
time.sleep(1)

bt.write(b"AT")
time.sleep(0.3)

print("RX:", bt.read(64))
```
